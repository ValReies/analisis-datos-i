##**Ficha de la fuente**

**Nombre de la fuente:** Banco Mundial - Indicadores de Desarrollo Mundial (World Bank Open Data API).

**Descripción:** Base de datos global que recopila indicadores macroeconómicos y sociales normalizados para todos los países miembros.

**Acceso:** API REST pública. Descarga verificada, sin token, devuelve JSON limpio.
**Cita:** Banco Mundial. (2026). World Development Indicators. Recuperado de la API oficial.

**Actualización:** Datos actualizados al 13 de julio de 2026.

**1. Quién levanta el dato y con qué propósito:**
El indicador es publicado por el Banco Mundial, pero recopilado a partir de fuentes oficiales nacionales (en el caso de México, datos del INEGI y Banco de México). El propósito del Banco Mundial al unificar estos datos es monitorear el desarrollo económico global, ayudar en el diseño de políticas públicas y evaluar las economías a nivel internacional de forma estandarizada.

**2. Cobertura:**
Periodo: Series de tiempo anuales (la API extrae los últimos años disponibles, generalmente desde 1960 a la actualidad).
Geografía y universo: Nivel país (México). Contempla la producción total formal de bienes y servicios dentro de las fronteras.
Qué queda fuera: La economía informal y el trabajo no remunerado (tareas domésticas, cuidado), ya que el PIB no captura estas transacciones.

**3. Unidad de observación:**
Cada renglón de nuestra tabla representa el reporte de un país en un año específico (Ej. México en el año 2023).

**4. Cómo se llega al dato publicado:**
Es una estimación agregada. Se calcula mediante el Sistema de Cuentas Nacionales del país. El Banco Mundial toma esa suma en moneda local y la convierte a dólares estadounidenses (USD corrientes) usando el tipo de cambio oficial de mercado, permitiendo comparar a México con el resto del mundo.

**5. Licencia y cita:**

### Escala de medicion de las variables
pais (Código ISO del país): Cualitativa nominal. Son etiquetas o categorías que no tienen un orden matemático ni jerarquía.

anio (Año del registro): Cuantitativa discreta / Escala de intervalo. Son números enteros que representan periodos cronológicos de tiempo.

pib_usd (Valor del PIB en USD): Cuantitativa continua / Escala de razón. Son números que pueden tener decimales y poseen un cero absoluto real (donde cero significaría total ausencia de producción).

###Las cuatro preguntas

**¿Qué pregunta busca responder esta investigación?**
Esta investigación busca responder cuál ha sido la evolución, la trayectoria de crecimiento y la volatilidad del Producto Interno Bruto (PIB) de México a lo largo de las últimas décadas.

**¿Por qué es relevante esta pregunta para los Negocios Internacionales?**
Este análisis es fundamental porque el PIB es un indicador clave del atractivo de inversión y la salud del mercado interno. Una economía estable atrae la Inversión Extranjera Directa y permite a las empresas transnacionales evaluar la viabilidad de expansión o proyectar la demanda. Además, al analizar los datos en dólares, los especialistas pueden medir la vulnerabilidad del país frente a crisis globales y devaluaciones históricas.

**¿Qué datos específicos se analizarán?**
Se analizarán los registros anuales de la economía de México extraídos de los Indicadores del Desarrollo Mundial mediante la API oficial del Banco Mundial. Específicamente, se utilizarán tres variables: el país de observación medido en escala cualitativa nominal, el año de registro cronológico en escala cuantitativa de intervalo, y el valor total de la producción formal en dólares actuales, que corresponde a una escala cuantitativa continua y de razón.

**¿Cuáles son los límites de esta investigación?**
Los límites principales radican en que el cálculo oficial del PIB excluye la economía informal y el trabajo no remunerado. Además, al utilizar dólares corrientes, las cifras pueden sufrir distorsiones por la inflación o fluctuaciones en el tipo de cambio, sin reflejar necesariamente cambios en la producción real. Finalmente, este indicador global no muestra cómo se distribuye la riqueza ni refleja la calidad de vida o la desigualdad social en el país.
