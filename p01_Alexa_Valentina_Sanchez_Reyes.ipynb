El argumento mapping de la función **ggplot**( ) define cómo se asignan las variables del conjunto de datos a las propiedades visuales del gráfico. ### Graficas de punto

Cada observacion se dibuja como un punto, y los puntos que comparten valor se apilan. La altura dde la pila es la frecuencia de esa categoria.

**Cuando usarla:** con pocos casos, cuanod interesa que se vea que detras de cada unidad hay una observacion. Con muchos casos deja de leerse  y conviene la barra.

```r
# grafica de puntos

ggplot(data = grupos, mapping = aes(x = carrera)) +
  geom_dotplot(binaxis = "x", stackdir = "up", dotsize = 0.5, fill = azul)
```

```r
ggplot(data = grupos, mapping = aes(x = str_wrap(carrera, 18))) +
  geom_dotplot(binaxis = "x", stackdir = "up", dotsize = 0.5, fill = azul)
scale_y_continuous(NULL, breaks = NULL)
```

```r
ggplot(data = grupos, mapping = aes(x = str_wrap(carrera, 18))) +
  geom_dotplot(binaxis = "x", stackdir = "up", dotsize = 0.5, fill = azul) +
  scale_y_continuous(NULL, breaks = NULL) +
  labs(title = "Un punto por cada integrante del grupo", x = NULL)
```

### Grafica de barras

Representa la frecuencia de cada categoria con la longitud de una barra. Todas las barreras arrrancan del mismo punto, asi que elojo compara los extremos.

**Cuando usarala:** para una variable categorica, siempre. Es la opcion por defecto y la mas dificl de leer con precision.

Se construye por capas. Primero la basica:

```r
ggplot(data = grupos, mapping = aes(x = carrera)) +
  geom_bar()
```

```r
# Mejora de grafica de barras
ggplot(data = grupos, mapping = aes(y = carrera)) +
  geom_bar(fill = azul) +
  labs(x = "Estudiantes", y = NULL)
```

#### Ordenar las categorias
ggplot ordena alfabeticamente, que casi nunca es el orden que interesa. Para ordenar por frecuencia se convierte la variable en factor.

Un factor es una variable categorica con sus niveles declarados y en un orden definido. fct_infreq() ordena por frecuencias;

fct_rev() invierte.

```r
ggplot(data = grupos, mapping = aes(y = fct_rev (fct_infreq(carrera)))) +
  geom_bar(fill =azul) +
  labs(x = "Estudiantes", y = NULL)
```

### Etiquetar valores

Cuando importa la cifra exacta, se escribe sobre la barra y se puede quitar el eje, que ya no aporta.

```r
ggplot(data = grupos, mapping = aes(y = fct_rev (fct_infreq(carrera)))) +
  geom_bar(fill = azul) +
  geom_text(stat = "count", aes(label = after_stat(count)),
           hjust = 1.4, # posicion horizontal de la etiqueta
           color = "white",
           size = 4
           ) +
labs(x = "Estudiantes", y = NULL)
```
